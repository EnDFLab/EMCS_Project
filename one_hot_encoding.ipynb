**Adding Day of week Column**

```python
def add_day_of_week(url,name):
  #df = pd.read_csv("https://raw.githubusercontent.com/EnDFLab/EMCS_Project/refs/heads/smart_meter_data_analytics/physics_preprocessed.csv")
  df = pd.read_csv(url)
  df["timestamp"] = pd.to_datetime(df["timestamp"])
  df.head()
  # Replace values greater than 1 in a specific column with 0

  #To make sunday from 1 and saturday 7 the operation %7+1 is carried out
  #eg 1%7 + 1 = 1+1 = 2
  #2%7 + 1 = 2+1 = 3
  #7%7 +1 = 0+1 = 1

  day_of_week = {"Sunday":1,
                "Monday":2,
                "Tuesday":3,
                "Wednesday":4,
                "Thursday":5,
                "Friday":6,
                "Saturday":7}

  for day in day_of_week:
    df[day] = (df["timestamp"].dt.dayofweek + 1) % 7 + 1
    df[day] = df[day].where(df[day] == day_of_week[day], 0)
    df[day] = df[day].where(df[day] < 1, 1) #Do nothing if value is zero else change to 1.
  #print(df[df["weekday"]==0])
  print(df.head())
  df.drop(columns=["Saturday"],inplace=True)
  return df

```

**Adding Month Column**

```python
# Define a dictionary for months
def add_month(df,name):
  months = {
      "January": 1,
      "February": 2,
      "March": 3,
      "April": 4,
      "May": 5,
      "June": 6,
      "July": 7,
      "August": 8,
      "September": 9,
      "October": 10,
      "November": 11,
      "December": 12,
  }

  # Iterate over the months and create a column for each
  for month in months:
      df[month] = df["timestamp"].dt.month
      df[month] = df[month].where(df[month] == months[month], 0)
      df[month] = df[month].where(df[month] < 1, 1)
  df.drop(columns=["December"],inplace=True)
  df.to_csv(name + "_with_one_hot_encoding.csv", index=False)
  return

```

```python
building = ["physics","biotech","boyshostel","civil","electrical","management","transformer"]

for id in building:
  url = "https://raw.githubusercontent.com/EnDFLab/EMCS_Project/refs/heads/smart_meter_data_analytics/"+id+"_preprocessed.csv"
  df_with_week = add_day_of_week(url,id)
  add_month(df_with_week,id)
```
